Array/List Implementation of Stack

```python
# declare a class
class Stack:
  def __init__(self):
    self.st = []

  # step to add the element in stack
  def push(self, x):
    self.st.append(x)

  # step to pop the element in stack implemented using list
  def pop(self):
    if len(self.st) == 0:
      return -1

    x = self.st[-1]
    self.st.pop()
    return x

  # step to fetch the top element
  def top(self):
    if len(self.st) == 0:
      return -1
    return self.st[-1]

  # step to find the size of the stack
  def size(self):
    return len(self.st)
```

Implementation of Stack using LinkedList

```python
class Node:
  def __init__(self, data):
    self.data = data
    self.next = None

class Stack:
  def __init__(self):
    self.top = None
    self.length = 0

  def push(self, x):
    self.length += 1
    if self.top is None:
      self.top = Node(x)
      return
    else:
      newNode = Node(x)
      newNode.next = self.top
      self.top = newNode

    def pop(self):
      if self.top is None:
        return -1
      x = self.top.data
      self.length -= 1
      self.top = self.top.next
      return x

    def getTop(self):
      return self.top.data

    def size(self):
      return self.length
```

Implementation of Queue using List/Array

```python
# declare a class
class Queue:
  def __init__(self):
    self.q = []
    self.front = -1

  # step to add the element in queue
  def push(self, x):
    self.st.append(x)
    if self.front == -1:
      self.front = 0

  # step to pop the element in queue implemented using list
  def pop(self):
    if len(self.q) == 0:
      return -1

    x = self.q[self.front]
    self.front += 1
    if self.front == len(self.q):
      self.front = -1
      self.q = []
    return x

  # step to fetch the top element
  def getFront(self):
    if len(self.q) == 0:
      return -1
    return self.q[self.front]

  # step to find the size of the queue
  def size(self):
    if self.front == -1:
      return 0
    return len(self.q) - self.front
```
