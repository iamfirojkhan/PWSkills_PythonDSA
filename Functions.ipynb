Global Variables and Local Variables

```python
n = int(786) #this is global variable

def scope():
  n = int(100) #This is local variable
  print(n)

scope()
print(n)
```

TYPES OF ARGUMENTS

```python
'''DEFAULT ARGUMENTS'''

def greet(message="Hi ", name= "User"):
  print(message, name)

greet("Hello! ")
```

```python
'''KEYWORD ARGUMENTS'''

def summary(name, age, message):
  print(message, "your name is", name, "and age is",age)

summary(message = "Hi!", age = 23, name = "Hello World")
```

```python
'''POSITIONAL ARGUMENTS'''

def position(x, y):
  print("x", x)
  print("y", y)

position(5, 6)
```

```python
'''VARIABLE ARGUMENTS'''

def sumation(*args):
  print(type(args))
  print(args)

  sum = 0
  for num in args:
    sum +=num

  return sum

sumation(1, 2, 3, 4, 5,6)
```

```python
#Can we pass normal parameters in variable argument

def sumvar(a, b, *args):
  print(a)
  print(b)
  print(*args)

sumvar(2, 3, 4, 5, 6, 7, 8, 9)
```

```python
def sumvar(*args,a, b):
  print(a)
  print(b)
  print(*args)

sumvar(2, 3, 4, 5, 6, 7, 8, 9)
```

```python
def sumvar(*args,a, b):
  print(a)
  print(b)
  print(*args)

sumvar(2, 3, 4, 5, 6, 7, a = 8, b = 9)
```

**KWARGS IN PYTHON**

```python
def func(**kwargs):
  print(type(kwargs))
  print(kwargs)

  for key, value in kwargs.items():
    print(key, "->", value)

func(name = "Hello", age = 23, City = "Bangalore")
```

```python
def func1(a, b, *args, **kwargs):
  print(a)
  print(b)
  print(args)
  print(kwargs)

func1(1, 2, 3, 4, 5, 6, 7, name = "Hello", favfood = "Biryani")
```

NESTED FUNCTIONS

```python
def Outer():
  print("Hi! I am from the outter function")

  def inner():
    print("Hi! I am from the inner function")

  return inner

fn = Outer()

fn()
```

In the upper section fn->Outer() function is called and print statement executes below it, there is inner function but is not being called as we need to add inner() to call the function, But while calling the outer function we called it as fn = Outer(), we have return as inner also, so fn will contain inner. fn becomes inner function and if we call the fn(), we will get the inner function print result

```python
#Pass By Value

num = 786

def modify_number(num):
  num += 1
  print("Number after modification", num)

modify_number(num)
print("Original number after modification", num)
```

```python
#Pass by Reference

my_list = [1, 2, 3, 4]

def modify_list(li):
  li.append(5)

print("Original list is", my_list)
modify_list(my_list)
print("Updated list is", my_list)
```
