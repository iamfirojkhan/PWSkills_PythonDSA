```python
#RECURSION
#->Base Case
#-> Recursive Case

def printnnumbers(i, n):

  #base case
  if i > n:
    return

  #recurive case
  print(i, end = " ")

  printnnumbers(i+1, n)

printnnumbers(1, 10)
```

```python
#Factorial of a number using recursion

def factorial(n):

  #baseCase
  if n == 0:
    return 1

  return n * factorial(n - 1)

print(factorial(5))
```

```python
#Recurive Stack

def func(n):
  if n == 0:
    return

  func(n - 1)
  print(n, end = " ")

func(3)
```

EUCLIDEAN ALGORITHM

```python
def gcd(a, b):
  if b == 0:
    return a

  return gcd(b, a % b)

gcd (5, 10)
```

LCM OF THE NUMBER

```python
def lcm(a, b):
  return a * b//gcd(b, a % b)

lcm( 4, 10)
```
