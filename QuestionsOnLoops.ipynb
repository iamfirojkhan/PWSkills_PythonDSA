WAP in Python to print sum of all even numbers between 1 to 20 using for loop

```python
sum = 0

for num in range(21):
  if num % 2 == 0:
    sum += num

print("Sum of all even number between 1 to 20 is", sum)
```

Write a program in python to print the total no of vowel in the given string

```python
str = input("Enter the string\n")
count = 0

for ch in str:
  if ch.lower() in ['a', 'e', 'i', 'o', 'u']:
    count +=1

print("Total no. of vowel present in the string is", count)
```
