```python
set1 = {1, 1, 1, "hello", "world", 2, 3, 5, 7, 1, 2, 5}

print(set1)
print(type(set1))
print(len(set1))
```

```python
list1 = [1, 2, 3, 1, 2, 3, 4, 8, 54, 90, 87, 786, 1, 54, 90]

# count the total number of unique elements in list

set2 = set(list1)
print(set2)
print(len(set2))
```

```python
set2 = {1, 2, 3, 4, 8, 786, 54, 87, 90}

set2.add(100)
print(set2)

# if we try to add the existing value then
set2.add(786)
print(set2)
```

```python
set2 = {1, 2, 3, 4, 8, 786, 54, 87, 90}

set2.discard(8)

# discarding the value which is not present in set using discard
set2.discard(77)

set2.remove(2)
# discarding the value which is not present in set using remove-> this will give us the error
set2.remove(77)
```

```python
set2 = {1, 2, 3, "hello", 4, "hii", 8, 786, 54, 87, 90}

# we can check if key is present in the set by using 'in'-> membership

if 'hello' in set2:
  print(True)
else:
  print(False)

if 'world' in set2:
  print(True)
else:
  print(False)
```

```python
# DEEP COPY

set1 = {1, 2, 3, 4, 5}
set2 = set1

set2.add(6)

print(set2)
print(set1)
```

```python
# SHALLOW COPY
set1 = {1, 2, 3, 4, 5}
set2 = set1.copy()

set2.add(6)

print(set2)
print(set1)
```

```python
set1 = {1, 2, 3, 4, 5, 6, 7, 8}
set2 = {2, 4, 6, 8}

# union
print(set1 | set2)
print(set1.union(set2))
```

```python
set1 = {1, 2, 3, 4, 5, 6, 7, 8}
set2 = {2, 4, 6, 8}

# intersection
print(set1 & set2)
print(set1.intersection(set2))
```

```python
set1 = {1, 2, 3, 4, 5, 6, 7, 8}
set2 = {2, 4, 6, 8}

#difference
print(set1 - set2)
print(set1.difference(set2))

print(set2 - set1)
print(set2.difference(set1))
```

```python
set1 = {1, 2, 3, 4, 5, 6, 7, 8}
set2 = {2, 4, 6, 8}

#symmetricdifference
print(set1 ^ set2)
print(set1.symmetric_difference(set2))
```

DICTIONARY

```python
dict1 = {}
print(type(dict1))
```

```python
dict1 = {1:"hello", "world":2, 10.5:"hii", "key":True}
print(type(dict1))
print(dict1)
```

```python
dict1 = {1:"hello", "world":2, 10.5:"hii", "key":True}
print(dict1[1])
```

```python
dict2 = {1:"Sultan", 2:"Menon", 3:"Raj", 4:"Yash"}

dict2[5] = "Masud"
print(dict2)
dict2[4] = "Idrish"
print(dict2)

dict2.update({1:"Salman", 6:"Sulaiman"})
print(dict2)
```

```python
dict3 = {1: 'Salman', 2: 'Menon', 3: 'Raj', 4: 'Idrish', 5: 'Masud', 6: 'Sulaiman'}

dict3.pop(2)
print(dict3)

dict3.get(3)
```

```python
dict4 = {1: 'Salman', 2: 'Menon', 3: 'Raj', 4: 'Idrish', 5: 'Masud', 6: 'Sulaiman'}

print(dict4.items()) #.items() return the key value pair into the tuples form

for key, values in dict4.items():
  print(key)

print(dict4.keys())
for key, values in dict4.items():
  print(values)

print(dict4.values())
for key, values in dict4.items():
  print(key, values)
```

```python
names = ["hello", "world", "day", "hello", "is", "world", "fine"]

freq = {}
# we will use dictionary to find the total frequency of the elements in the list

for name in names:
  if name not in freq:
    freq[name] = 1
  else:
    freq[name] += 1

print(freq)
```

```python
from collections import defaultdict
freq = defaultdict(list)

print(freq)
```
